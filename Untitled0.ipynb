```python
password = "Syedsasdf"

upper = False
lower = False
digit = False
n = len(password)
for c in password:
  if c.isupper():
    upper = True
  elif c.islower():
    lower = True
  elif c.isdigit():
    digit = True

if n>=8 and upper==True and lower==True and digit==True:
  print("strong password")
else:
  print("weak password")
```
